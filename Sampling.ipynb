```python
def get_wdl_dict(G):
    nodes = list(G.nodes())
    wdl_dict = dict()
    
    for u in nodes:
        wdl_u = u.split(":")[-1]
        if wdl_u != "-2":
            continue
    
        wdl_u_count = 0
        for v in G.successors(u):
            wdl_v = v.split(":")[-1]
            if wdl_v == wdl_u:
                wdl_u_count += 1
    
        #for v in G.predecessors(u):
        #    wdl_v = v.split(":")[-1]
        #    if wdl_v == wdl_u:
        #        wdl_u_count += 1
                
        wdl_dict[str(u)] = wdl_u_count
    
    return {k: v for k, v in sorted(wdl_dict.items(), key=lambda item: -item[1])}
```

```python
G_wdl_dict = get_wdl_dict(G)
```

```python
G_wdl_dict_ = get_wdl_dict(G_)
```

```python
for k in G_wdl_dict_.keys():
    print(f"{k}, original: {G_wdl_dict[k]}, subgraph: {G_wdl_dict_[k]}")
```
